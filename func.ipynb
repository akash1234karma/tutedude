```python
def evenorodd(num):
    if num % 2 == 0:
        print("num is even")
    elif num % 2 !=0:
        print("num is odd")
    else:
        type(num) == str
        

```

```python
num = int(input("Enter any number:"))
evenorodd(num)
```

```python
def eveodd(num):
    if isinstance(num,int):
        if num%2==0:
            print("even")
        else:
            print("odd")
    elif isinstance(num,str):
        print("Invalid")
    else:
        print(f"{num} is invalid")
```

```python
print(eveodd(False))
```

```python
def factorial(num):
    if num == 0 or num == 1:
        print(f"Factorial {num} is 1")
    else:
        factorial = 1
        for i in range(1, num+1):
            factorial=factorial*i
        print(f"factorial is num:{factorial}")
        
```

```python
number=int(input("Enter number"))
factorial(number)
```

```python
def fact_recursive(num):
    if num==0 or num==1:
        return 1
    else:
        return num*fact_recursive(num-1)
```

```python
fact_recursive(0)
```

```python
a = int(input("enter num"))
print(f"{a}, {fact_recursive(a)}")
```

```python
def var(*position, **keyword):
    for i in position:
        print(f'position:{position}')
    for key, value in keyword.items():
        print(f'{key} {value}')
        
```
