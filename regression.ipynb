```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X=data.drop(columns=['cognitive_load'])
```

```python
Y=data['cognitive_load']
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours','commits', 'bugs_reported', 'ai_usage_hours']]=scale.fit_transform(X_train[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours','commits', 'bugs_reported', 'ai_usage_hours']])
X_test[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours','commits', 'bugs_reported', 'ai_usage_hours']]=scale.fit_transform(X_test[['hours_coding', 'coffee_intake_mg', 'distractions', 'sleep_hours','commits', 'bugs_reported', 'ai_usage_hours']])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential([Dense(64,input_shape=(X_train.shape[1],)), Dense(1,activation='linear')])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',metrics=['mae'], loss='mse')
```

```python
model.fit(X_train,Y_train, epochs=30)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
Y_predict=model.predict(X_test)
```

```python
mean_squared_error(Y_predict,Y_test)
```
