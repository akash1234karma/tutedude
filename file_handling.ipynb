```python
def count_lines(file):
    with open(file,'r') as files:
        lines=files.readines()
        Total_lines=len(lines)
        Total_words=sum(len(line.split()for line in lines))
        Total_char=sum(len(line) for  line in lines)
    return Total_lines, Total_words, Total_char

file_path = "Total_data.txt"

Total_lines,Total_char,Total_words=count_lines(file_path)
print(f'{Total_char} {Total_lines} {Total_words}')
```
