```python
import torch
```

```python
input_tokens=tokenizer('Write a python code for adding two number', return_tensors='pt')
```

```python
# Step 1
from transformers import AutoTokenizer, AutoModelForCausalLM

# Step 2
model_name = "Qwen/Qwen3-0.6B"

# Step 3
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Step 4
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    dtype="float32"
)

print("Model loaded successfully")
```

```python
out=model(input_ids=input_tokens['input_ids'])
```

```python
gen_out=model.generate(input_ids=input_tokens['input_ids'], max_new_tokens=100)
gen_out
```

```python
print(tokenizer.batch_decode(gen_out))
```
