```python
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

message_hist=ChatMessageHistory()

message_hist.add_user_message('Hello')
message_hist.add_ai_message('Hi there !')
print(message_hist)
```

```python
store={}
```

```python
## Purpose is this function in each session that user info is going to have history
def get_history(session_id: str):
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]
```

```python
llm=ChatOpenAI(model='gpt-3.5-turbo')
prompt=ChatPromptTemplate.from_messages([('system','you are helpful AI'),('human','{input}')])
parser=StrOutputParser()
chain = prompt | llm | parser

chain_with_history=RunnableWithMessageHistory(chain,get_history,input_messages_key='input',output_messages_key='output')

response1=chain_with_history.invoke({'input':'My name is akash karma'}, config={'configurable':{'session_id':'amit123'}})
response2=chain_with_history.invoke({'input':'what is my name'}, config={'configurable':{'session_id':'amit123'}})

```
