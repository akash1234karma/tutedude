```python
from youtube_transcript_api import YouTubeTranscriptApi
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
load_dotenv()
```

```python
yt_api=YouTubeTranscriptApi()
```

```python
video_id='v0ynfDPpe4E'
transcript=yt_api.fetch(video_id)
```

```python
transcript
```

```python
transcript_text=" ".join(doc.text for doc in transcript)
```

```python
transcript_text
```

```python
splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
```

```python
chunks=splitter.split_text(transcript_text)
```

```python
chunks
```

```python
len(chunks)
```

```python
embeddings=OpenAIEmbeddings(model='text-embedding-3-large')
```

```python
vector_store=FAISS.from_texts(chunks,embeddings)
```

```python
vector_store
```

```python
vector_store.similarity_search("what is RAG")
```

```python
retriever=vector_store.as_retriever(search_kwargs={"k":4})
```

```python
retriever.invoke("what is RAG pipeline?")
```

```python
prompt=PromptTemplate(template="""you are a helpful assistant, Answer from the context, if context is insufficient then directly say I don't know {context} Question {question}""",
               input_variable=['context', 'question'])
```

```python
question= 'what is Deep learning??'
```

```python
context=retriever.invoke(question)
```

```python
context=" ".join(i.page_content for i in context)
```

```python
final=prompt.invoke({'question':question,"context":context})
```

```python
llm=ChatOpenAI(model='gpt-3.5-turbo')
```

```python
ans=llm.invoke(final)
```

```python
print(ans.content)
```
