```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory
from dotenv import load_dotenv
load_dotenv()

llm=ChatOpenAI(model='gpt-3.5-turbo')
parser=StrOutputParser()
prompt=ChatPromptTemplate.from_messages([('system','you are friendly chatbot, you have access to past conversation history'),
                                  MessagesPlaceholder(variable_name='history'),('human','{input}')])
chain = prompt | llm | parser

store={}
chat=ChatMessageHistory()
def get_history(session_id:str):
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]

chain_with_hisory=RunnableWithMessageHistory(chain,get_history,input_messages_key='input',history_messages_key='history',output_messages_key='output')

response1=chain_with_hisory.invoke({'input':"what is my name?"},config={'configurable':{'session_id':'amit123'}})
response1
```
