```python
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
load_dotenv()
```

```python
google_model = ChatGoogleGenerativeAI(model='gemini-2.5-flash')
gpt_model=ChatOpenAI(model='gpt-3.5-turbo')
```

```python
prompt1=PromptTemplate(template='Generate a detailed report on the {Topic}', input_variables=['Topic'])
prompt2=PromptTemplate(template='Create a 3 points summary on {Report}', input_variables=['Report'])
```

```python
parser=StrOutputParser()
```

```python
chain = prompt1 | google_model | parser | prompt2 |gpt_model | parser
```

```python
response=chain.invoke({'Topic':'What is RAG Evaluation'})
```
