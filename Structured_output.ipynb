```python
from pydantic import BaseModel, Field
from typing import Optional, Literal
from langchain_openai import ChatOpenAI
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
class Data(BaseModel):
    Product : str = Field(description="Provide the name of the product")
    Pros: list[str] = Field(description='Mention positive points mentioned about product')
    Cons: list[str] = Field(description='Mention negative points mentioned about product')
    Sentiment: Literal['Positive','Negative']= Field(description='Mnetioned the sentiment of the user review')
    Ratings: float=Field(ge=1,le=5,default=3,description='Return rating of the reviewer of current product')
    Reviewer: Optional[str]=Field(description='Name of the reviewer', default=None)
```

```python
Review = """I recently bought the Noise Cancelling Headphones 500 and I've been using them daily for the past two weeks. 
The sound quality is absolutely fantastic, with deep bass and clear vocals. The noise cancellation works very well, especially on flights, 
and the battery easily lasts more than 18 hours on a single charge. However, I found them a bit uncomfortable after wearing them for more than 3 hours continuously, 
as the ear cups get slightly warm. Also, the companion mobile app is not very intuitive. Overall, I would rate them 4.5 out of 5 because of the amazing performance and long battery life, 
but with room for improvement in comfort and app design."""
```

```python
model=ChatOpenAI(model='gpt-3.5-turbo')
```

```python
structured_respone=model.with_structured_output(Data)
```

```python
response=structured_respone.invoke(Review)
```
