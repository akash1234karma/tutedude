```python
import pandas as pd 
```

```python
data=pd.read_csv('huge_1M_titanic.csv')
```

```python
data.info()
```

```python
data.columns
```

```python
data=data.drop(columns=['PassengerId', 'Name','Age','Ticket','Cabin'])
```

```python
data['Embarked']=data['Embarked'].replace({'S':'Southamtom','C':'Cheboung','Q':'Queensland'})
```

```python
data.columns
```

```python
data.info()
```

```python
data.sample()
```

```python
data.dropna(subset=['Embarked'], inplace=True)
```

```python
data.columns
```

```python
data.sample()
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
label=LabelEncoder()
ohe=OneHotEncoder(sparse_output=False)
```

```python
data['Sex']=label.fit_transform(data['Sex'])
```

```python
Embarked=ohe.fit_transform(data[['Embarked']])
```

```python
Embarked
```

```python
Embarked=pd.DataFrame(Embarked,columns=ohe.get_feature_names_out())
```

```python
Embarked
```

```python
data.sample()
```

```python
data=pd.concat([data.drop(columns=['Embarked']),Embarked],axis=1)
```

```python
data.sample()
```

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
```

```python
data.columns
```

```python
num_cols=['Pclass','SibSp',	'Parch','Fare']
```

```python
data[num_cols]=scale.fit_transform(data[num_cols])
```

```python
import pickle

with open('label_encoder.pkl', 'wb') as file:
    pickle.dump(label,file)
with open('onehot_encoder.pkl', 'wb') as file:
    pickle.dump(ohe,file)
with open('scale.pkl','wb') as file:
    pickle.dump(scale,file)
 

```

```python
data.isnull().sum()
```

```python
data=data.dropna()
```

```python
data.isnull().sum()
```

```python
#Train validatiom and test

X=data.drop(columns='Survived')
Y=data['Survived']
```

```python
#Model selection

from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=20, random_state=42)
```

```python
X_train,X_valid,Y_train,Y_valid=train_test_split(X_train,Y_train,test_size=0.2, random_state=42)
```

```python
import tensorflow
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping


```

```python
(X_train.shape[1],)
```

```python
model=Sequential([Dense(128,input_shape=(X_train.shape[1],),activation='relu'),
            Dense(64,activation='relu'),
            Dense(32,activation='relu'),Dense(1,activation='sigmoid')])
```

```python
model.summary()
```

```python
#Model compliation optimizer, loss, metrics
```

```python
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.01)
```

```python
tensorflow.keras.losses.binary_crossentropy
```

```python
model.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
stopping_callback=EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

```python
model.fit(X_train,Y_train,validation_data=(X_valid,Y_valid), epochs=3,callbacks=[stopping_callback])
```

```python
model.save('model.h5')
```
