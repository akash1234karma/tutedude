```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
```

```python
with open('label_encoder.pkl', 'rb') as file:
    label=pickle.load(file)
with open('onehot_encoder.pkl', 'rb') as file:
    ohe=pickle.load(file)
with open('scale.pkl', 'rb') as file:
    scale=pickle.load(file)
```

```python
model = load_model('model.h5')
```

```python
model.summary()
```

```python
user={'Pclass':1,'Sex':'male','SibSp':2,'Parch':1,'Fare':76,'Embarked':'Cheboung'}
```

```python
user = pd.DataFrame([user])
```

```python
user
```

```python
user['Sex']=label.transform(user['Sex'])
```

```python
embarked=ohe.transform(user[['Embarked']])
```

```python
embarked=pd.DataFrame(embarked, columns=ohe.get_feature_names_out())
```

```python
user=pd.concat([user.drop(columns=['Embarked']),embarked],axis=1)
```

```python
user
```

```python
user[['Pclass','SibSp','Parch','Fare']]=scale.transform(user[['Pclass','SibSp','Parch','Fare']])
```

```python
user
```

```python
y=model.predict(user)
```

```python
y
```

```python
y[0][0]
```

```python
if y > 0.5:
    print('survive')
else:
    print('not')
```
