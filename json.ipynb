```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()

model = ChatOpenAI(model='gpt-3.5-turbo')
parser=JsonOutputParser()
prompt=PromptTemplate(template='Provide me the with name, age, city of the fiction person and {specific_format}',
            input_variables=[], partial_variables={'specific_format':parser.get_format_instructions()})

'''final_prompt=prompt.format()
response=model.invoke(final_prompt)
print(response.content)'''

#OR

chain = prompt | model | parser
result=chain.invoke({})
print(result)
```

```python
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field

class Data(BaseModel):
    name:str=Field(description='return the name of the user')
    age:int=Field(description='return the age of user')
    city:str=Field(description='the city has to be from Madhya Pradesg state')
    state:str=Field(description='State has to Madhyan Pradesh')

parser=PydanticOutputParser(pydantic_object=Data)
prompt=PromptTemplate(template='Provide the name,city,age and state of fiction person from {nationality} in {format}',
               input_variables=['nationality'],partial_variables={'format':parser.get_format_instructions()})
'''final_prompt=prompt.invoke({'nationality':'Indian'})
response=model.invoke(final_prompt)
parser.parse(response.content)'''

#OR BY CHAIN

chain = prompt | model | parser
result=chain.invoke({'nationality':'Indian'})
result
```
