```python
import ast
```

```python
# ast.literal_eval() is a func used got safely parse abd evaluate a string containig a pythom lib
```

```python
def genre_extract(col):
    L=[]
    for i in ast.literal_eval(col):
        L.append(i['name'])
    return L
```

```python
data['genres']=data['genres'].apply(genre_extract)
```

```python
def cast_extract(col):
    L=[]
    j=1
    for i in ast.literal_eval(col):
        if j<=3:
            L.append(i['name'])
            j+=1
    return L
```

```python
data['cast']=data['cast'].apply(cast_extract)
```

```python
def crew_extract(col):
    L=[]
    for i in ast.literal_eval(col):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L
```

```python
data['crew']=data['crew'].apply(crew_extract)
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps=PorterStemmer()
```

```python
def stemming(text):
    L=[]
    for i in text.split():
        L.append(ps.stem(i))
    return " ".join(L)
```

```python
data['tags']=data['tags'].apply(stemming)
```

```python
data['tags']
```

```python
#Stop word removal
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors=cv.fit_transform(data['tags'])
```

```python
vectors
```

```python
vectors=vectors.toarray()
```

```python
vectors
```

```python
cv
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarities=cosine_similarity(vectors)
```

```python
similarities
```

```python
data[data['title']=='avatar'].index[0]
```

```python
def recommend(movie):
    movie_index=data[data['title']==movie].index[0]
    recommendations=similarities[movie_index]
    movie_list=sorted(enumerate(recommendations),reverse=True,key=lambda x:x[1])[1:6]
    for i in movie_list:
        print(data.iloc[i[0]].title)
```

```python
recommend('jaws')
```

```python
recommend('lol')
```
