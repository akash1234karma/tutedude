```python
import pandas as pd
```

```python
data=pd.read_json('News_Category_Dataset_v3.json',lines=True)
```

```python
data.columns
```

```python
data=data['short_description'].head(2000)
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense
```

```python
tokenizer=Tokenizer(oov_token='Nothing')
```

```python
tokenizer.fit_on_texts(data)
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
data=tokenizer.texts_to_sequences(data)
```

```python
data
```

```python
input_sequences=[]

for seq in data:
    if len(seq)<2:
        continue
    for i in range(1,len(seq)):
        ngrams=seq[:i+1]
        input_sequences.append(ngrams)
```

```python
input_sequences
```

```python
max_len=max([len(i) for i in input_sequences])
```

```python
max_len
```

```python
input=pad_sequences(input_sequences,padding='pre',maxlen=max_len)
```

```python
input
```

```python
X=input[: :-1]
Y=input[:,-1]
```

```python
Y=np.array(Y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
vocab_size=len(tokenizer.word_index)+1
```

```python
max_len=X.shape[1]
```

```python
model=Sequential([Embedding(vocab_size,128,input_length=max_len),LSTM(256),Dense(vocab_size,activation='softmax')])
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.build(input_shape=(None,max_len))
```

```python
model.summary()
```

```python
model.fit(X_train,Y_train,epochs=1, validation_data=(X_test,Y_test))
```

```python
model.save('nextword_model.h5')
```
