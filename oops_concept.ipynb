```python
class Cartoon:

    def __init__(self, name, show, phrase):
        self.name=name
        self.show=show
        self.phrase=phrase

    def introduction(self):
        print(f'hello my name is {self.name}')
    def speak(self):
        print(f'{self.name} says {self.phrase}')
        
```

```python
mickey=Cartoon("mickey","show", "hello")
```

```python
mickey.introduction()
```

```python
mickey.speak()
```

```python
class Avengers:
    @staticmethod
    def mission():
        print("Protect earth")

    origin="Avenger"
    def __init__(self,name,power):
        self.name=name
        self.power=power

    @classmethod
    def origin_info(cls):
        print(f"{self.origin}")

    def intro(self):
        print(f'{self.name}')
```
