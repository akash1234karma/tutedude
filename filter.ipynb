```python
def even_func(value):
    if value%2==0:
        return True
```

```python
iterable_list = [1,2,3,232,33,22,3,2323,2323,23,23,2132,43,9,44,434,43,543,536,56,456,5]

print(list(filter(even_func, iterable_list)))
```

```python
print(list(filter(lambda x:x**2, iterable_list)))
```

```python
print(list(filter(lambda x:x>50 and x%2==0, iterable_list)))
```

```python
employees = [{'name':'jay', 'age': 32, 'gender':'male'},{'name':'raj', 'age': 23, 'gender':'male'},{'name':'soni', 'age': 32, 'gender':'female'}]
```

```python
def age_greater(employee):
    if employee['age'] >= 32:
        return True
```

```python
print(list(filter(age_greater,employees)))
```
