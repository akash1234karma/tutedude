```python
from langchain_community.embeddings import OpenAIEmbeddings
```

```python
embeddings=OpenAIEmbeddings(model='text-embedding-3-large')
```

```python
query = 'hello i am you new teacher'
```

```python
embeddings.embed_query(query)
```

```python
from langchain_chroma import Chroma
```

```python
from langchain_core.documents import Document
```

```python
docs=[Document(page_content='SRK is a famous action in bollywood', metadata={'actor':'Shahrukh khan', 'identity':'King'}),
      Document(page_content='Amitabh is BigB in bollywood', metadata={'actor':'Amitabh Bachan', 'identity':'Bigb'})]
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
vector_store=Chroma(embedding_function=OpenAIEmbeddings(),persist_directory='chroma_database',collection_name='simple_data')
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=['documents','embeddings','metadatas'])
```

```python
vector_store.similarity_search(query='Amit',k=2 )
```

```python
vector_store.similarity_search_with_score(query='Amit',k=2 )
```

```python
update_srk_doc=Document(page_content='Shahrukh khan is new movie Pathan', metadata={'actor':'Shahrukh'})
```

```python
vector_store.update_document(document_id='a33db4b3-bc3d-4145-9bc3-dd782d85b734',document=update_srk_doc)
```

```python
vector_store.get(include=['documents','embeddings','metadatas'])
```

```python
vector_store.delete(ids=['86eb26e9-9d6a-4301-89c9-5d15e43e4e95'])
```

```python
vector_store.get(include=['documents','embeddings','metadatas'])
```

```python
from langchain_community.vectorstores import FAISS
```

```python
db=FAISS.from_documents(docs,embeddings)
```

```python
db.similarity_search(query='who is Khan', k=1)
```

```python
db.similarity_search_with_score(query='who is Khan',k=1)
```

```python
db.save_local('faiss_indexex')
```

```python
new_db=db.load_local('faiss_indexex',embeddings,allow_dangerous_deserialization=True)
```

```python
new_db.similarity_search(query='king',k=1)
```
