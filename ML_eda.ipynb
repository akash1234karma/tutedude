```python
import pandas as pd
import seaborn as sns
```

```python
test=pd.read_csv('tested.csv')
```

```python
test.shape
```

```python
test1=pd.read_csv('tested.csv')
```

```python
data=pd.concat([test,test1])
```

```python
data.shape
```

```python
data['Family']=data['SibSp'] + data['Parch'] + 1
```

```python
sns.barplot(data=data, x='Family', y='Survived')
```

```python
def fam_type(num):
    if num>=1 and num<=3:
        return 'Small Family'
    elif num>=3 and num<=5:
        return 'Medium family'
    else:
        return 'Large family'
```

```python
data['Family']=data['Family'].apply(fam_type)
```

```python
data
```

```python
data['Family'].value_counts()
```

```python
pd.crosstab(data['Family'].values, data['Survived'].values, normalize=True)
```
