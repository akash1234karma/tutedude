```python
class Avenger:

    def __init__(self, name, weapon, power, nationality='American'):
        self.name=name
        self.weapon=weapon
        self.power=power
        self.nationality=nationality

    def introduction(self):
        print(f'hello ny name is {self.name}')
    def carry_weapon(self):
        print(f'i am carrying a {self.weapon}')
        
```

```python
iron_name=Avenger("Tony","repulsor","genious")
print(iron_name.introduction())
print(iron_name.carry_weapon())
```

```python
captain=Avenger("steve","shield","strong","india")
print(captain.introduction())
print(captain.carry_weapon())
```
