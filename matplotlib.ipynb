```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sales = pd.read_csv('Chocolate1.csv')
```

```python
sales.info()
```

```python
sales.sample(4)
```

```python
sales['Date']=pd.to_datetime(sales['Date'], format='%d-%m-%Y')
sales['Date']=sales['Date'].dt.strftime('%b')
month_order=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
sales['Date']=pd.Categorical(sales['Date'], categories=month_order, ordered=True)
```

```python
sales.sample(3)
```

```python
data=sales.groupby('Date')['Amount'].sum()
```

```python
data
```

```python
plt.plot(data.index,data.values)
plt.xlabel('Month')
plt.ylabel('Amount')
plt.title("Month vs Amount")
plt.show()
```
