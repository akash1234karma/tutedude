```python
class Animal:
    def __init__(self, name,sound):
        self.name=name
        self.sound=sound
    #getter

    def sound_info(self):
        print(f'the {self.name} sound is {self.sound}')

    #setter

    def sound_modify(self, new_sound):
        self.new_sound=new_sound
        print(f'{self.new_sound} is the new sound of {self.name}')
        
```

```python
animal=Animal("lion","roar")
```

```python
animal.sound_info()
```

```python
animal.sound_modify("meow")
```

```python
class Grandparent:
    def __init__(self, name, locker_key):
        self.name=name
        self.locker_key=locker_key

    def locker_info(self):
        print(f'The locker details for {self.name} is {self.locker_key}')
        
```

```python
class Parent(Grandparent):
    def __init__(self, name, locker_key):
        super().__init__(name, locker_key)

    def child_reveal(self):
        print(f'the locker key info is {self.locker_key}')
```

```python
grandparent=Grandparent("AMIT", 2423)
```

```python
grandparent.locker_info()
```

```python
parent=Parent("amit", 2323)
```

```python
parent.child_reveal()
```

```python
class Bankaccount:
    def __init__(self, name,balance):
        self.name=name
        self.balance=balance

    def get_balance(self):
        print(f'the total balance is {self.balance}')

    def deposit(self, amount):
        if amount >0:
            self.balance += amount

            return f'deposit amount is : {self.get_balance()}'
        else:
            return "deposit amount is negtive"

    def withdraw(self, withdraw):
        if 0<withdraw<=self.balance:
            self.balance = withdraw
            return f"withdraw {withdraw} is {self.get_balance()}"
        else:
            "Insufficient" 
```

```python
bank=Bankaccount("amit", 1000)
```

```python
bank.get_balance()
```

```python
bank.withdraw(100)
```

```python
bank.get_balance()
```
