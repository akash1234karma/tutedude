```python
class Avenger:
    def __init__(self, name, strength):
        self.name=name
        self.strength=strength

    def __add__(self, other):
        return Avenger(self.name + "&" + other.name, (self.strength + other.strength)/2)
    def __sub__(self, other):
        return Avenger(self.name,self.strength - other.strength)
    def __mul__(self, other):
        return Avenger(self.name,self.strength * other.strength)
    def __truediv__(self, other):
        return Avenger(self.name,self.strength / other.strength)
    def __str__(self):
        return f'{self.name} with power {self.strength}'
```

```python
thor = Avenger("thor", 100)
hulk = Avenger("hulk", 190)
```

```python
team = thor + hulk
```

```python
print(team)
```

```python
class AIModel:
    def __init__(self,name,accuracy):
        self.name=name
        self.accuracy=accuracy

    def __gt__(self, other):
        return self.accuracy > other.accuracy
    def __lt__(self,other):
        return self.accuracy < other.accuracy
    def __eq__(self, other):
        return self.accuracy == other.accuracy

```

```python
gpt=AIModel("gpt-3", 100)
bard=AIModel("Google", 98)
```

```python
gpt > bard
```
