```python
class Avenger:

    def __init__(self,name,power):
        self.name=name
        self.power=power
    def intro(self):
        print(f'{self.name}')

class Ironman(Avenger):

    def __init__(self, name, power, weapon):
        super().__init__(name, power)
        self.weapon=weapon

    def weapon_details(self):
        print(f'{self.weapon}')
```

```python
Iron=Ironman('tony','repulsor','nothing')
```

```python
Iron.weapon_details()
```

```python
class Grandfather:

    def __init__(self,Nationality,Bald,anger):
        self.Nationality=Nationality
        self.Bald=Bald
        self.anger=anger

    def intro(self):
        print(f'{self.Nationality} and {self.anger}')

class Parent(Grandfather):

    def __init__(self, Nationality, Bald, anger, intelligence):
        super().__init__(Nationality, Bald, anger)
        self.intelligence=intelligence

    def intelligence_details(self):
        print(f'{self.intelligence}')

class Child(Parent):
    def __init__(self, Nationality, Bald, anger, intelligence,charm):
        super().__init__(Nationality, Bald, anger, intelligence)
        self.charm=charm

    def charm_details(self):
        print(f'i am {self.charm}')
```

```python
parent = Parent("Indian","no","lil","yes")
```

```python
parent.intelligence
```

```python
parent.intro()
```

```python
child=Child("India","NO","NO","NO","NO")
```

```python
child.intelligence_details()
```

```python
class Ironman:

    def __init__(self,name,power,suit):
        self.name=name
        self.power=power
        self.suit=suit

    def intro(self):
        print(f'{self.name} and {self.power}')

class CaptainAmerica:

    def __init__(self, name, power, suit):
        self.name=name
        self.power=power
        self.suit=suit

    def suit_detail(self):
        print(f'{self.suit}')

class Hulk(Ironman,CaptainAmerica):
    def __init__(self, name, power, suit, angerissue):
        super().__init__(name, power, suit)

        self.angerissue=angerissue
    def angerdesc(self):
        print(f'{self.angerissue}')
```

```python
test=Hulk("banner","strong","nothing","high")
```

```python
test.suit_detail()
```

```python
test.intro()
```

```python
test.angerdesc()
```
