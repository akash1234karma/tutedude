```python
class Grandfather:
    def speak(self):
        print("i am amit")
class Parent(Grandfather):
    def speak(self):
        print("i am amit's son")
class Child(Grandfather):
    def speak(self):
        print("i am amit's grandson")
```

```python
def attribute_provider(objects):
    print(objects.speak())

```

```python
grandfather=Grandfather()
```

```python
attribute_provider(grandfather)
```

```python
parent=Parent()
```

```python
attribute_provider(parent)
```
