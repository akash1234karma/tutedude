```python
class Grandfather:
    def speak(self):
        print("i am amit")
class Parent(Grandfather):
    def speak(self):
        print("i am amit's son")
class Child(Grandfather):
    def speak(self):
        print("i am amit's grandson")
```

```python
def attribute_provider(objects):
    print(objects.speak())

```

```python
grandfather=Grandfather()
```

```python
attribute_provider(grandfather)
```

```python
parent=Parent()
```

```python
attribute_provider(parent)
```

```python
class shape:
    def area(self):
        return "This func the area"
```

```python
class Rectangle(shape):
    def __init__(self,height,width):
        self.height=height
        self.width=width
    def area(self):
        return self.width * self.height
```

```python
class Circle(shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14*self.radius*self.radius
```

```python
shape=shape()
```

```python
shape.area()
```

```python
rect=Rectangle(3,5)
```

```python
rect.area()
```

```python
circle=Circle(5)
```

```python
circle.area()
```
