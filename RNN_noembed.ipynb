```python
data.rename(columns={'Positive':'Sentiment','i am getting on borderlands and i will murder you all':'Tweets'}, inplace=True)
```

```python
data=data[(data['Sentiment']=='Positive')| (data['Sentiment']=='Negative')|(data['Sentiment']=='Neutral')]
```

```python
data
```

```python
data['Sentiment'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
label=LabelEncoder()
```

```python
data['Sentiment']=label.fit_transform(data['Sentiment'])
```

```python
data.sample()
```

```python
import keras
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer(oov_token='Nothing')
```

```python
tokenizer.fit_on_texts(data['Tweets'])
```

```python
data.sample()
```

```python
tokenizer.word_index
```

```python
tokenizer.document_count
```

```python
len(tokenizer.word_index)
```

```python
sequences=tokenizer.texts_to_sequences(data['Tweets'])
```

```python
sequences
```

```python
len(sequences[2])
```

```python
from keras.utils import pad_sequences
```

```python
sequences=pad_sequences(sequences, padding='pre')
```

```python
sequences
```

```python
from tensorflow.keras.layers import SimpleRNN, Dense,Input, Embedding
from tensorflow.keras.models import Sequential
```

```python
X=sequences
Y=data['Sentiment'].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2, random_state=42)
```

```python
model=Sequential()
```

```python
model.add(SimpleRNN(32,input_shape=(99,1),return_sequences=False))
model.add(Dense(3,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',metrics=['accuracy'],loss='sparse_categorical_crossentropy')
```

```python
model.fit(X_train,Y_train,epochs=10,validation_data=(X_test,Y_test),batch_size=64)
```
