```python
import pandas as pd
```

```python
data=pd.read_csv('twitter_training.csv')
```

```python
data=data.sample(10000,random_state=42)
```

```python
data=data.dropna()
```

```python
data.drop(columns=['2401','Borderlands'],inplace=True)
```

```python
data.sample()
```

```python
data.rename(columns={'Positive':'Sentiment','im getting on borderlands and i will murder you all ,':'Tweets'},inplace=True)
```

```python
data.sample()
```

```python
data=data[(data['Sentiment']=='Positive')|(data['Sentiment']=='Negative')|(data['Sentiment']=='Neutral')]
```

```python
data.sample()
```

```python
from sklearn.preprocessing import LabelEncoder
label=LabelEncoder()
```

```python
data['Sentiment']=label.fit_transform(data['Sentiment'])
```

```python
import tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer(oov_token='Nothing')
```

```python
tokenizer.fit_on_texts(data['Tweets'])
```

```python
tokenizer.word_index
```

```python
tokenizer.word_counts
```

```python
tokenizer.document_count
```

```python
len(tokenizer.word_index)
```

```python
sequences=tokenizer.texts_to_sequences(data['Tweets'])
```

```python
sequences[2
          ]
```

```python
from keras.utils import pad_sequences
```

```python
sequences=pad_sequences(sequences,padding='pre')
```

```python
sequences
```

```python
from tensorflow.keras.layers import SimpleRNN,Dense,Input,Embedding
from tensorflow.keras.models import Sequential
```

```python
X=sequences
Y=data['Sentiment'].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
model=Sequential()
```

```python
model.add(SimpleRNN(32,input_shape=(99,1), return_sequences=False))
model.add(Dense(3,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',metrics=['accuracy'],loss='sparse_categorical_crossentropy')
```

```python
model.fit(X_train,Y_train,epochs=5, validation_data=(X_test,Y_test), batch_size=64)
```

```python
vocab_size=len(tokenizer.word_index)+1
```

```python
model_2=Sequential()
model_2.add(Input(shape=(99,1)))
model_2.add(Embedding(input_dim=vocab_size,output_dim=64))
model_2.add(Dense(3,activation='softmax'))
```

```python
model_2.summary()
```

```python
model_2.compile(optimizer='adam',metrics=['accuracy'],loss='sparse_categorical_crossentropy')
```

```python
model_2.fit(X_train,Y_train,epochs=3,validation_data=(X_test,Y_test),batch_size=64)
```

```python
model_final=Sequential([Embedding(input_dim=vocab_size,output_dim=128,input_length=99,mask_zero=True),SimpleRNN(32),Dense(3,activation='softmax')])
```

```python
model_final.compile(optimizer='adam',metrics=['accuracy'],loss='sparse_categorical_crossentropy')
```

```python
model_final.summary()
```

```python
model_final.fit(X_train,Y_train,epochs=3,validation_data=(X_test,Y_test),batch_size=32)
```

```python
import pickle

with open('tokenizer.pkl','wb') as file:
    pickle.dump(tokenizer,file)

model_final.save('model.h5')
```
