```python
a = int(input("Enter a number:"))
b = int(input("Enter another number:"))
operation = input("Enter the operator (+,-,*,/)")

if operation == "+":
    print (a+b)
elif operation == "-":
    print(a-b)
elif operation == "*":
    print(a*b)
elif operation == "/":
    if b!=0:
        print(a/b)
    else:
        print("cannot divide by zero")
else:
    print("Invalid opertor")
```
