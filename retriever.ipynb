```python
from langchain_community.vectorstores import FAISS
```

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embeddings=OpenAIEmbeddings()
```

```python
docs=["Solar energy is a clean and renewable source of energy obtained from sunlight."
    "Solar panels capture sunlight and convert it into electricity, which can be used in homes, offices, industries, and other facilities."
    "One of the main advantages of solar energy is that sunlight is naturally available and renewable." 
    "Using solar power can reduce dependence on fossil fuels such as coal, oil, and natural gas. It can also help reduce greenhouse gas emissions associated with electricity generation."
"Solar energy systems can be installed on rooftops or developed as large solar farms. Although installation costs and dependence on available sunlight can be challenges, solar technology continues to play an important role in the transition toward cleaner and more sustainable energy sources."]
```

```python
vector_store=FAISS.from_texts(docs,embeddings)
```

```python
retriever=vector_store.as_retriever(search_type='mmr',search_kwargs={"k":2,"lambda_mult":0.5})
```

```python
query='what is solar panel?'
```

```python
result=retriever.invoke(query)
```

```python
print("MMR result")
for i in result:
    print("-",i.page_content)
```

```python
from langchain_classic.retrievers import MultiQueryRetriever
```

```python
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
embeddings=OpenAIEmbeddings()
LLM=ChatOpenAI(model='gpt-3.5-turbo')
```

```python
docs = [
    "Deep Learning uses neural networks with many layers to learn complex patterns.",
    "Generative AI uses Deep Learning models like GANs and Transformers to create new content.",
    "The breakthrough in Generative AI came from large-scale training of deep neural networks.",
    "Transformers, a Deep Learning architecture, enabled large language models like GPT.",
    "The Eiffel Tower is located in Paris, France.",
    "Photosynthesis is the process by which plants make food using sunlight.",
    "Python is a popular programming language for web development and data analysis."
]
```

```python
vector_store=FAISS.from_texts(docs,embeddings)
```

```python
multi_retriever=MultiQueryRetriever.from_llm(retriever=vector_store.as_retriever(),llm=LLM)
```

```python
query='what are the important 5 libraries in python to learn AI?'
```

```python
result=multi_retriever.invoke(query)
```

```python
print("\n Multiquery result: ")
for i in result:
    print("-",i.page_content)
```

```python
from langchain_classic.retrievers import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
```

```python
query='how do i reset my router passord ?'
```

```python
docs=["""The """]
```

```python
vector_store=FAISS.from_texts(docs,embeddings)
```

```python
LLM=ChatOpenAI(model='gpt-3.5-turbo')
```

```python
base_retriever=vector_store.as_retriever()
```

```python
compressor=LLMChainExtractor.from_llm(LLM)
```

```python
compressor_retriever=ContextualCompressionRetriever(base_compressor=compressor, base_retriever=base_retriever)
```

```python
result=compressor_retriever.invoke(query)
```

```python
print("\n Contextual compressor result:")
for i in result:
    print("-",i.page_content)
```
